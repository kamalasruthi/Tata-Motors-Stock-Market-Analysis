```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv(
    "../data/processed/tata_motors_cleaned.csv",
    index_col=0,
    parse_dates=True
)

os.makedirs("../outputs/charts", exist_ok=True)

sns.set_theme(style="whitegrid")

print("Dataset ready for advanced analysis!")
```

```python
Step 2: Monthly returns heatmap
```

```python
# Calculate monthly closing prices
monthly_prices = df["adj_close"].resample("ME").last()

# Calculate monthly percentage returns
monthly_returns = monthly_prices.pct_change() * 100

# Create year-month table
monthly_table = monthly_returns.to_frame("return")

monthly_table["year"] = monthly_table.index.year
monthly_table["month"] = monthly_table.index.month

heatmap_data = monthly_table.pivot(
    index="year",
    columns="month",
    values="return"
)

# Month labels
heatmap_data.columns = [
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug",
    "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(15, 9))

sns.heatmap(
    heatmap_data,
    cmap="RdYlGn",
    center=0,
    annot=True,
    fmt=".1f",
    linewidths=0.5
)

plt.title("Tata Motors Monthly Returns Heatmap (2000–2025)")
plt.xlabel("Month")
plt.ylabel("Year")

plt.tight_layout()

plt.savefig(
    "../outputs/charts/monthly_returns_heatmap.png",
    dpi=300
)

plt.show()
```

```python
Step 3: Stock volatility analysis
```

```python
# Calculate daily returns
daily_returns = df["adj_close"].pct_change()

# Calculate 30-day rolling volatility
rolling_volatility = (
    daily_returns
    .rolling(window=30)
    .std() * np.sqrt(252) * 100
)

plt.figure(figsize=(15, 6))

plt.plot(
    rolling_volatility.index,
    rolling_volatility,
    label="30-Day Annualized Volatility"
)

plt.title("Tata Motors Historical Volatility")
plt.xlabel("Year")
plt.ylabel("Annualized Volatility (%)")

plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    "../outputs/charts/volatility_analysis.png",
    dpi=300
)

plt.show()

print("Average annualized volatility:",
      rolling_volatility.mean(), "%")

print("Maximum annualized volatility:",
      rolling_volatility.max(), "%")
```

```python
Step 4: Maximum drawdown analysis
```

```python
# Calculate running maximum price
running_peak = df["adj_close"].cummax()

# Calculate drawdown
drawdown = (
    (df["adj_close"] - running_peak)
    / running_peak
) * 100

# Maximum drawdown
max_drawdown = drawdown.min()

# Date of maximum drawdown
max_drawdown_date = drawdown.idxmin()

print("Maximum Drawdown:", round(max_drawdown, 2), "%")
print("Drawdown Date:", max_drawdown_date)

# Visualize drawdown
plt.figure(figsize=(15, 6))

plt.fill_between(
    drawdown.index,
    drawdown,
    0,
    alpha=0.4
)

plt.plot(drawdown.index, drawdown)

plt.title("Tata Motors Maximum Drawdown Analysis")
plt.xlabel("Year")
plt.ylabel("Drawdown (%)")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    "../outputs/charts/drawdown_analysis.png",
    dpi=300
)

plt.show()
```

```python
Step 5: Correlation analysis
```

```python
correlation_columns = [
    "open",
    "high",
    "low",
    "close",
    "adj_close",
    "volume",
    "daily_return"
]

correlation_matrix = df[correlation_columns].corr()

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Tata Motors Stock Data Correlation Matrix")

plt.tight_layout()

plt.savefig(
    "../outputs/charts/correlation_matrix.png",
    dpi=300
)

plt.show()
```

```python
Step 6: Annual risk and return analysis
```

```python
# Daily adjusted returns
returns = df["adj_close"].pct_change()

# Annual returns using year-end adjusted closing prices
year_end_prices = df["adj_close"].resample("YE").last()

annual_returns = year_end_prices.pct_change() * 100

# Annualized volatility for each year
annual_volatility = (
    returns.groupby(returns.index.year)
    .std() * np.sqrt(252) * 100
)

# Combine results
risk_return = pd.DataFrame({
    "Annual Return": annual_returns.values,
    "Annualized Volatility": annual_volatility.values
}, index=annual_volatility.index)

risk_return.index.name = "Year"

print(risk_return.round(2))

# Scatter plot
plt.figure(figsize=(11, 7))

plt.scatter(
    risk_return["Annualized Volatility"],
    risk_return["Annual Return"],
    alpha=0.7
)

for year, row in risk_return.dropna().iterrows():
    plt.annotate(
        str(year),
        (row["Annualized Volatility"], row["Annual Return"]),
        fontsize=8
    )

plt.axhline(0, linewidth=1)
plt.title("Tata Motors Annual Risk vs Return")
plt.xlabel("Annualized Volatility (%)")
plt.ylabel("Annual Return (%)")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.savefig(
    "../outputs/charts/risk_return_analysis.png",
    dpi=300
)

plt.show()
```

```python
Step 7: Generate an analytical summary
```

```python
# Overall performance
first_price = df["adj_close"].iloc[0]
last_price = df["adj_close"].iloc[-1]

total_return = (
    (last_price / first_price) - 1
) * 100

# Daily return statistics
average_daily_return = df["daily_return"].mean()
best_day = df["daily_return"].max()
worst_day = df["daily_return"].min()

# Annual performance
annual_returns_clean = annual_returns.dropna()

best_year = annual_returns_clean.idxmax()
worst_year = annual_returns_clean.idxmin()

# Maximum drawdown
max_drawdown = drawdown.min()

# Summary
print("=" * 55)
print("TATA MOTORS STOCK MARKET ANALYSIS - KEY FINDINGS")
print("=" * 55)

print(f"First adjusted price: ₹{first_price:.2f}")
print(f"Last adjusted price: ₹{last_price:.2f}")

print(f"\nOverall price return: {total_return:.2f}%")

print(f"Average daily return: {average_daily_return:.4f}%")
print(f"Best daily return: {best_day:.2f}%")
print(f"Worst daily return: {worst_day:.2f}%")

print(f"\nBest performing year: {best_year}")
print(f"Worst performing year: {worst_year}")

print(f"\nMaximum drawdown: {max_drawdown:.2f}%")

print("=" * 55)
```
